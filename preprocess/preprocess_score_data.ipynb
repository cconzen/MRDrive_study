### Restructuring Order Group

Data samples are currently equipped with a flat order_group attribute.

f.e.

| participant_id | order_group                         |
|----------------|-------------------------------------|
| 01             | G1 - Lumo - Coda - Nelo + A - B - C |

Restructuring dataframe to be shown in separate rows:

| block_id | participant_id | group | position | conditon | route |
|----------|----------------|-------|----------|----------|-------|
| 1_1      | 01             | G1    | 1        | Lumo     | A     |
| 1_2      | 01             | G1    | 2        | Coda     | B     |
| 1_3      | 01             | G1    | 3        | Nelo     | C     |



```python
all_rows = []
stsad_abbr = ["Trust", "Performance", "NDRT", "Risk", "Judgement", "Reaction"]

#Order Group =  "G1 - Lumo - Coda - Nelo + A - B - C"
for idx, row in df_scores.iterrows():
    # split at '+' -> "G1 - Lumo - Coda - Nelo" , "A - B - C"
    left, right = row['order_group'].split(' + ')
    # split left into "G1" and ["Lumo", "Coda", "Nelo"]
    group, *conditions = [x.strip() for x in left.split("-")]
    # split right at "-" -> ["A", "B", "C"]
    routes = [x.strip() for x in right.split("-")]

    rows = []

    for pos, (cond, route) in enumerate(zip(conditions, routes), start=1):
        # compute STSaD index range for this block
        start = (pos - 1) * 6 + 1
        stsad_values = {
            f"stsad_{stsad_abbr[j-1]}": row[f"stsad_{start + j - 1}"]
            for j in range(1, 7)
        }

        new_row = {
            "block_id": f"{row['participant_id']}_{pos}",
            "participant_id": row['participant_id'],
            "group": group,
            "position": pos,
            "condition": cond,
            "route": route,
            **stsad_values,
            f"nasa_mental_demand": row[f"nasa_mental_demand_{pos}"],
            f"nasa_effort": row[f"nasa_effort_{pos}"],
            f"nasa_frustration": row[f"nasa_frustration_{pos}"],
        }

        print(new_row)
        all_rows.append(new_row)

df_scores_split = pd.DataFrame(all_rows)
```

## System Understanding Scoring


```python
scoring_key = {
    "event1":  {"correct": "event1_1",  "partial": "event1_2",  "distractors": ["event1_3", "event1_4"]},
    "event2":  {"correct": "event2_1",  "partial": "event2_2",  "distractors": ["event2_3", "event2_4"]},
    "event3":  {"correct": "event3_1",  "partial": "event3_2",  "distractors": ["event3_3", "event3_4"]},
    "event4":  {"correct": "event4_1",  "partial": "event4_2",  "distractors": ["event4_3", "event4_4"]},
    "event5":  {"correct": "event5_1",  "partial": "event5_2",  "distractors": ["event5_3", "event5_4"]},
    "event6":  {"correct": "event6_1",  "partial": "event6_2",  "distractors": ["event6_3", "event6_4"]},
    "event7":  {"correct": "event7_1",  "partial": "event7_2",  "distractors": ["event7_3", "event7_4"]},
    "event8":  {"correct": "event8_1",  "partial": "event8_2",  "distractors": ["event8_3", "event8_4"]},
    "event9":  {"correct": "event9_1",  "partial": "event9_2",  "distractors": ["event9_3", "event9_4"]},
    "event10": {"correct": "event10_1", "partial": "event10_2", "distractors": ["event10_3", "event10_4"]},
    "event11": {"correct": "event11_1", "partial": "event11_2", "distractors": ["event11_3", "event11_4"]},
    "event12": {"correct": "event12_1", "partial": "event12_2", "distractors": ["event12_3", "event12_4"]},
}

def reverse_score(series, likert_max=7):
    return (likert_max + 1) - series

def cronbach_alpha(df_items):
    """
    Compute Cronbach's alpha for a dataframe of items.
    Rows = participants, columns = items
    """
    df_items = df_items.dropna()
    k = df_items.shape[1]

    if k < 2:
        return np.nan

    item_variances = df_items.var(axis=0, ddof=1)
    total_score = df_items.sum(axis=1)
    total_variance = total_score.var(ddof=1)

    if total_variance == 0:
        return np.nan

    alpha = (k / (k - 1)) * (1 - item_variances.sum() / total_variance)
    return alpha


def score_system_understanding(df, scoring_key, partial_weight=0.5, likert_max=7):
    """
    Adds scored columns to a copy of the dataframe:
    - one score per event
    - overall system understanding score

    Formula:
    (correct + partial_weight*partial + reverse(d1) + reverse(d2)) / (1 + partial_weight + n_distractors)
    """
    df_scored = df.copy()
    event_score_cols = []

    for event, key in scoring_key.items():
        correct_col = key["correct"]
        partial_col = key["partial"]
        distractor_cols = key["distractors"]

        correct = df_scored[correct_col]
        partial = df_scored[partial_col] * partial_weight
        reversed_distractors = sum(reverse_score(df_scored[col], likert_max=likert_max) for col in distractor_cols)

        denominator = 1 + partial_weight + len(distractor_cols)

        df_scored[f"{event}_score"] = (correct + partial + reversed_distractors) / denominator
        event_score_cols.append(f"{event}_score")

    df_scored["system_understanding_score"] = df_scored[event_score_cols].mean(axis=1)

    return df_scored, event_score_cols


def create_scored_items(df, scoring_key, partial_weight=0.5, likert_max=7):
    """
    Creates a dataframe where every item is transformed in the direction
    of better understanding:
    - correct = raw
    - partial = weighted raw
    - distractors = reverse-scored
    """
    scored_items = pd.DataFrame(index=df.index)

    for event, key in scoring_key.items():
        scored_items[f"{event}_correct"] = df[key["correct"]]
        scored_items[f"{event}_partial"] = df[key["partial"]] * partial_weight

        for i, col in enumerate(key["distractors"], start=1):
            scored_items[f"{event}_distractor{i}_rev"] = reverse_score(df[col], likert_max=likert_max)

    return scored_items


##################


df = df_scores
df_scored, event_score_cols = score_system_understanding(df, scoring_key)
scored_items = create_scored_items(df, scoring_key)

# Cronbachs alpha across all transformed items
alpha_all_items = cronbach_alpha(scored_items)

# Cronbachs alpha across event scores
alpha_event_scores = cronbach_alpha(df_scored[event_score_cols])

print("Event score columns:", event_score_cols)
print("Cronbach's alpha (all scored items):", round(alpha_all_items, 3))
print("Cronbach's alpha (event scores):", round(alpha_event_scores, 3))


# df_scored.to_csv("system_understanding_scored.csv", index=False)
```

```python
all_rows = []
stsad_abbr = ["Trust", "Performance", "NDRT", "Risk", "Judgement", "Reaction"]

route_event_map = {
    "A": ["event1_score", "event2_score", "event3_score", "event4_score"],
    "B": ["event5_score", "event6_score", "event7_score", "event8_score"],
    "C": ["event9_score", "event10_score", "event11_score", "event12_score"],
}

# Order Group = "G1 - Lumo - Coda - Nelo + A - B - C"
for idx, row in df_scored.iterrows():
    # split at '+'
    left, right = row['order_group'].split(' + ')

    # split left into group and conditions
    parts_left = [x.strip() for x in left.split("-")]
    group = parts_left[0]
    conditions = parts_left[1:]

    # split right into routes
    routes = [x.strip() for x in right.split("-")]

    for pos, (cond, route) in enumerate(zip(conditions, routes), start=1):
        # compute STSaD index range for this block
        start = (pos - 1) * 6 + 1
        stsad_values = {
            f"stsad_{stsad_abbr[j-1]}": row[f"stsad_{start + j - 1}"]
            for j in range(1, 7)
        }

        # get event score columns for this route
        event_cols = route_event_map[route]
        event_scores = {
            f"route_event_{i+1}_score": row[event_col]
            for i, event_col in enumerate(event_cols)
        }

        event_scores_named = {
            event_col: row[event_col]
            for event_col in event_cols
        }

        # route-level understanding score
        route_understanding_score = row[event_cols].mean()

        new_row = {
            "block_id": f"{row['participant_id']}_{pos}",
            "participant_id": row['participant_id'],
            "group": group,
            "position": pos,
            "condition": cond,
            "route": route,
            **stsad_values,
            "nasa_mental_demand": row[f"nasa_mental_demand_{pos}"],
            "nasa_effort": row[f"nasa_effort_{pos}"],
            "nasa_frustration": row[f"nasa_frustration_{pos}"],
            **event_scores,
            **event_scores_named,
            "route_system_understanding_score": route_understanding_score,
        }

        all_rows.append(new_row)

df_events_split = pd.DataFrame(all_rows)
```

```python
plot_df = df_events_split.dropna(subset=["condition", "route", "route_system_understanding_score"]).copy()

conditions = sorted(plot_df["condition"].unique())
routes = sorted(plot_df["route"].unique())

fig, ax = plt.subplots(figsize=(10, 6))

positions = []
data = []
labels = []

width = 0.22
gap = 1.0

for i, cond in enumerate(conditions):
    base = i * gap
    for j, route in enumerate(routes):
        vals = plot_df[
            (plot_df["condition"] == cond) &
            (plot_df["route"] == route)
        ]["route_system_understanding_score"].dropna()

        if len(vals) > 0:
            positions.append(base + j * width)
            data.append(vals)
            labels.append(f"{cond}\n{route}")

ax.boxplot(data, positions=positions, widths=0.18)

ax.set_xticks([i * gap + width for i in range(len(conditions))])
ax.set_xticklabels(conditions)
ax.set_ylabel("Route System Understanding Score")
ax.set_title("Route System Understanding by Condition and Route")

plt.tight_layout()
plt.show()
```
